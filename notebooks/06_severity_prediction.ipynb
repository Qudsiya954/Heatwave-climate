```python
import pandas as pd
import numpy as np

severity_df = pd.read_parquet(
    "../Data/processed/heatwave_severity.parquet1"
)

print(severity_df.shape)
print(severity_df.head())
print(severity_df.columns)
```

```python
print(severity_df["target_severity"].value_counts())
print(
    severity_df["target_severity"]
    .value_counts(normalize=True)
)
```

```python
print(severity_df.columns.tolist())
```

```python
import pandas as pd
import numpy as np

severity_df["date"] = pd.to_datetime(severity_df["date"])

target = "target_severity"

features = [
    col for col in severity_df.columns
    if col not in [
        "date",
        "target_severity",
        "severity_label"
    ]
]

X_train = severity_df.loc[
    severity_df["date"].dt.year <= 2022,
    features
]

y_train = severity_df.loc[
    severity_df["date"].dt.year <= 2022,
    target
]

X_val = severity_df.loc[
    severity_df["date"].dt.year == 2023,
    features
]

y_val = severity_df.loc[
    severity_df["date"].dt.year == 2023,
    target
]

X_test = severity_df.loc[
    severity_df["date"].dt.year >= 2024,
    features
]

y_test = severity_df.loc[
    severity_df["date"].dt.year >= 2024,
    target
]

print("Features:", len(features))
print("Train:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)
```

```python
label_map = {
    1.0: 0,
    2.0: 1,
    3.0: 2
}

y_train = y_train.map(label_map)
y_val = y_val.map(label_map)
y_test = y_test.map(label_map)

print(y_train.value_counts())
print(y_val.value_counts())
print(y_test.value_counts())
```

```python
from xgboost import XGBClassifier

xgb_severity = XGBClassifier(
    n_estimators=400,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,

    objective="multi:softprob",
    num_class=3,
    eval_metric="mlogloss",

    tree_method="hist",
    random_state=42,
    n_jobs=-1
)

xgb_severity.fit(
    X_train,
    y_train,
    eval_set=[(X_val, y_val)],
    verbose=False
)
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    f1_score
)

val_pred = xgb_severity.predict(X_val)

print("Accuracy:", accuracy_score(y_val, val_pred))
print("Macro F1:", f1_score(y_val, val_pred, average="macro"))
print("Weighted F1:", f1_score(y_val, val_pred, average="weighted"))

print("\nClassification Report:")
print(
    classification_report(
        y_val,
        val_pred,
        target_names=[
            "Moderate",
            "High",
            "Extreme"
        ]
    )
)

print("\nConfusion Matrix:")
print(confusion_matrix(y_val, val_pred))
```

```python
test_pred = xgb_severity.predict(X_test)

print("Accuracy:", accuracy_score(y_test, test_pred))
print("Macro F1:", f1_score(y_test, test_pred, average="macro"))
print("Weighted F1:", f1_score(y_test, test_pred, average="weighted"))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        test_pred,
        target_names=[
            "Moderate",
            "High",
            "Extreme"
        ]
    )
)

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, test_pred))
```

```python
importance = (
    pd.Series(
        xgb_severity.feature_importances_,
        index=features
    )
    .sort_values(ascending=False)
)

print(importance)
```

```python
import lightgbm as lgb

lgb_severity = lgb.LGBMClassifier(
    objective="multiclass",
    num_class=3,

    n_estimators=400,
    learning_rate=0.05,
    max_depth=-1,
    num_leaves=31,

    subsample=0.8,
    colsample_bytree=0.8,

    random_state=42,
    n_jobs=-1
)

lgb_severity.fit(
    X_train,
    y_train,
    eval_set=[(X_val, y_val)],
    callbacks=[
        lgb.early_stopping(50, verbose=False)
    ]
)
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    f1_score
)

val_pred_lgb = lgb_severity.predict(X_val)

print("Accuracy:", accuracy_score(y_val, val_pred_lgb))
print("Macro F1:", f1_score(y_val, val_pred_lgb, average="macro"))
print("Weighted F1:", f1_score(y_val, val_pred_lgb, average="weighted"))

print("\nClassification Report:")
print(
    classification_report(
        y_val,
        val_pred_lgb,
        target_names=[
            "Moderate",
            "High",
            "Extreme"
        ]
    )
)

print("\nConfusion Matrix:")
print(confusion_matrix(y_val, val_pred_lgb))
```

```python
test_pred_lgb = lgb_severity.predict(X_test)

print("Accuracy:", accuracy_score(y_test, test_pred_lgb))
print("Macro F1:", f1_score(y_test, test_pred_lgb, average="macro"))
print("Weighted F1:", f1_score(y_test, test_pred_lgb, average="weighted"))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        test_pred_lgb,
        target_names=[
            "Moderate",
            "High",
            "Extreme"
        ]
    )
)

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, test_pred_lgb))
```

```python
import pandas as pd
import numpy as np

regression_df = pd.read_parquet(
    "../Data/processed/temperature_regression.parquet"
)

regression_df["date"] = pd.to_datetime(regression_df["date"])

print(regression_df.shape)
```

```python
target = "target_temperature"

features = [
    col for col in regression_df.columns
    if col not in ["date", target]
]

train_mask = regression_df["date"].dt.year <= 2022
val_mask   = regression_df["date"].dt.year == 2023
test_mask  = regression_df["date"].dt.year >= 2024

X_train = regression_df.loc[train_mask, features]
y_train = regression_df.loc[train_mask, target]

X_val = regression_df.loc[val_mask, features]
y_val = regression_df.loc[val_mask, target]

X_test = regression_df.loc[test_mask, features]
y_test = regression_df.loc[test_mask, target]

print("Features:", len(features))
print("Train:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)
```

```python
import lightgbm as lgb

lgb_reg = lgb.LGBMRegressor(
    objective="regression",

    n_estimators=2000,
    learning_rate=0.03,

    num_leaves=63,
    max_depth=-1,

    min_child_samples=50,
    min_split_gain=0.0,

    subsample=0.85,
    subsample_freq=1,
    colsample_bytree=0.85,

    reg_alpha=0.1,
    reg_lambda=0.5,

    random_state=42,
    n_jobs=-1
)
```

```python
lgb_reg.fit(
    X_train,
    y_train,

    eval_set=[(X_train, y_train), (X_val, y_val)],
    eval_names=["train", "validation"],

    callbacks=[
        lgb.early_stopping(
            stopping_rounds=100,
            verbose=True
        ),
        lgb.log_evaluation(100)
    ]
)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

val_pred = lgb_reg.predict(
    X_val,
    num_iteration=lgb_reg.best_iteration_
)

mae = mean_absolute_error(y_val, val_pred)
rmse = np.sqrt(mean_squared_error(y_val, val_pred))
r2 = r2_score(y_val, val_pred)

print("Validation MAE :", mae)
print("Validation RMSE:", rmse)
print("Validation R²  :", r2)
```

```python
test_pred = lgb_reg.predict(
    X_test,
    num_iteration=lgb_reg.best_iteration_
)

test_mae = mean_absolute_error(y_test, test_pred)
test_rmse = np.sqrt(mean_squared_error(y_test, test_pred))
test_r2 = r2_score(y_test, test_pred)

print("Test MAE :", test_mae)
print("Test RMSE:", test_rmse)
print("Test R²  :", test_r2)
```

```python
importance = (
    pd.Series(
        lgb_reg.feature_importances_,
        index=features
    )
    .sort_values(ascending=False)
)

print(importance)
```
